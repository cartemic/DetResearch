```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

sns.set_context("notebook")
```

```python
lfs = pd.read_hdf("laminar_flame_speeds_2024_09_21.h5")
lfs.head()
```

```python
lfs["cj_ratio"] = lfs["lfs"] / lfs["cj"]
lfs["meas_ratio"] = lfs["lfs"] / lfs["wave_speed"]
for y in ("lfs", "cj_ratio", "meas_ratio"):
    sns.relplot(x="phi_nom", y=y, data=lfs, col="dil_mf_nom", hue="diluent")
    plt.suptitle(y, weight="bold")
    plt.subplots_adjust(top=0.875)
```
